### 4.1 Customer Demographic Attributes

| Attribute | Description | Data Type |
|---|---|---|
| `age` | Age of the customer | Numeric |
| `job` | Type of job | Categorical |
| `marital` | Marital status | Categorical |
| `education` | Education level | Categorical |

**Possible values:**

- `job`: admin., blue-collar, entrepreneur, housemaid, management, retired, self-employed, services, student, technician, unemployed, unknown.
- `marital`: divorced, married, single, unknown. The specification notes that divorced includes divorced or widowed.
- `education`: basic.4y, basic.6y, basic.9y, high.school, illiterate, professional.course, university.degree, unknown.

### 4.2 Credit and Loan Attributes

| Attribute | Description | Data Type |
|---|---|---|
| `default` | Indicates whether the customer has credit in default | Categorical |
| `housing` | Indicates whether the customer has a housing loan | Categorical |
| `loan` | Indicates whether the customer has a personal loan | Categorical |

Possible values for these attributes include `yes`, `no`, and `unknown`.

### 4.3 Current Campaign Contact Attributes

These variables describe the last contact with the customer during the current marketing campaign.

| Attribute | Description | Data Type |
|---|---|---|
| `contact` | Communication type: cellular or telephone | Categorical |
| `month` | Month of the last contact | Categorical |
| `day_of_week` | Day of the last contact | Categorical |
| `duration` | Duration of the last contact, measured in seconds | Numeric |

Possible values for `month` are January through December, represented by `jan`, `feb`, `mar`, and so on. The `day_of_week` variable contains `mon`, `tue`, `wed`, `thu`, and `fri`.

**Important consideration regarding `duration`:**

The project specification states that call duration strongly affects the target variable. For example, if duration is zero, the target is `no`.

However, call duration is not known before a call takes place. Once the call has ended, the outcome may already be known. Therefore, `duration` should be excluded from a realistic predictive model intended to identify potential subscribers before a call. It may be included separately for benchmark purposes.

### 4.4 Other Campaign Attributes

| Attribute | Description | Data Type |
|---|---|---|
| `campaign` | Number of contacts performed during the current campaign for the customer, including the last contact | Numeric |
| `pdays` | Number of days since the customer was last contacted during a previous campaign | Numeric |
| `previous` | Number of contacts performed before the current campaign for the customer | Numeric |
| `poutcome` | Outcome of the previous marketing campaign | Categorical |

Possible values for `poutcome` are `failure`, `nonexistent`, and `success`.

**Important consideration regarding `pdays`:** The value `999` indicates that the customer was not previously contacted.

### 4.5 Social and Economic Context Attributes

| Attribute | Description | Data Type |
|---|---|---|

| `cons.price.idx` | Consumer price index; monthly indicator | Numeric |
| `cons.conf.idx` | Consumer confidence index; monthly indicator | Numeric |
| `euribor3m` | Three-month Euribor rate; daily indicator | Numeric |
| `nr.employed` | Number of employees; quarterly indicator | Numeric |

These variables represent social and economic conditions associated with the campaign period.

### 4.6 Output Variable

| Attribute | Description | Data Type |
|---|---|---|
| `y` | Indicates whether the customer subscribed to a term deposit | Binary categorical |

The target variable has two possible values:

- `yes`: The customer subscribed to a term deposit.
- `no`: The customer did not subscribe to a term deposit.

The variable `y` will be used as the target for predictive modeling.

# 5. Data Preprocessing and Feature Preparation

## 5.1 Objective

The objective of this stage is to prepare the dataset for machine learning by handling duplicate records, identifying input features, separating the target variable, and preparing the data for model training.

The categorical `unknown` values will initially be retained as categories because they may contain useful information about unavailable customer details.

The `duration` variable will be excluded from the primary prediction model because it is not available before a marketing call takes place.

The dataset will be split into training and testing sets using stratified sampling to preserve the proportion of subscription outcomes.

```python
# Create a cleaned copy of the dataset

df_clean = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("Duplicate records removed:", len(df) - len(df_clean))
```

```python
# Define features and target

X = df_clean.drop(
    columns=["y", "duration", "previous_contact_status"]
)

y = df_clean["y"].map({
    "no": 0,
    "yes": 1
})

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Identify numerical and categorical features
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

# Preprocess numerical variables
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocess categorical variables
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
# Create the baseline classification pipeline

baseline_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

# Train the model
baseline_model.fit(X_train, y_train)

print("Baseline Logistic Regression model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Generate predictions
y_pred = baseline_model.predict(X_test)
y_prob = baseline_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, y_prob), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["No Subscription", "Subscription"]
))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Predicted No", "Predicted Yes"],
    yticklabels=["Actual No", "Actual Yes"]
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted Outcome")
plt.ylabel("Actual Outcome")
plt.tight_layout()
plt.show()
```

```python
from sklearn.base import clone
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)
from sklearn.pipeline import Pipeline
import pandas as pd

# Define models
models = {
    "Logistic Regression": LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ),
    "Decision Tree": DecisionTreeClassifier(
        class_weight="balanced",
        min_samples_leaf=20,
        random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        min_samples_leaf=5,
        n_jobs=-1,
        random_state=42
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=200,
        class_weight="balanced",
        min_samples_leaf=5,
        n_jobs=-1,
        random_state=42
    )
}

# Train and evaluate each model
results = []
trained_models = {}

for name, classifier in models.items():
    model = Pipeline(steps=[
        ("preprocessor", clone(preprocessor)),
        ("classifier", classifier)
    ])

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "ROC_AUC": roc_auc_score(y_test, probabilities)
    })

    trained_models[name] = model

# Display model comparison
model_comparison = (
    pd.DataFrame(results)
    .set_index("Model")
    .sort_values("ROC_AUC", ascending=False)
)

display(model_comparison.round(4))
```

### 7.7 Precision-Recall Curve Analysis

The Random Forest precision-recall curve illustrates how precision changes as recall increases.

Higher recall enables the bank to identify more actual subscribers but may increase the number of customers incorrectly predicted to subscribe. Higher precision produces a more selective list of potential subscribers but may miss some actual subscribers.

Therefore, the final classification threshold should be selected according to the bank's marketing capacity and the relative business costs of false-positive and false-negative predictions.

```python
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Use the already-trained Random Forest
rf_model = trained_models["Random Forest"]
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]

thresholds_to_test = [0.20, 0.30, 0.40, 0.50, 0.60]

threshold_results = []

for threshold in thresholds_to_test:
    predictions = (
        rf_probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test, predictions, labels=[0, 1]
    ).ravel()

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "Customers_Flagged": int(predictions.sum()),
        "True_Subscribers_Found": int(tp),
        "False_Positive_Contacts": int(fp),
        "Missed_Subscribers": int(fn)
    })

threshold_comparison = pd.DataFrame(threshold_results)

display(threshold_comparison.round(4))
```

### 7.8 Classification Threshold Analysis

The Random Forest model was evaluated at classification thresholds of 0.20, 0.30, 0.40, 0.50, and 0.60.

Lower thresholds increase recall by identifying more potential subscribers, but they also increase false-positive predictions. Higher thresholds improve precision while reducing the proportion of actual subscribers identified.

The threshold of 0.50 achieved a precision of 45.30%, recall of 61.21%, and F1-score of 52.06%. At a threshold of 0.60, precision increased to 49.19%, recall decreased to 55.39%, and the F1-score was 52.10%.

The threshold of 0.60 produced the highest F1-score among the tested thresholds, although its advantage over 0.50 was negligible.

A threshold of 0.50 will be retained as the baseline for further evaluation, with 0.60 considered as an alternative for a more selective marketing campaign. The final threshold should be selected using the bank's campaign capacity and the relative costs of false-positive and false-negative predictions.

# 10. Challenges Faced and Solutions

## 10.1 Class Imbalance

**Challenge:** The dataset contains 36,548 non-subscribers and 4,640 subscribers before duplicate removal. Approximately 11.27% of customers subscribed to the term deposit, resulting in an imbalanced target variable.

**Technique used:** Stratified train-test splitting and `class_weight="balanced"` for the classification models.

**Reason:** Stratification preserves the approximate proportion of subscribers and non-subscribers in the training and testing sets. Class weighting gives greater importance to the minority class during model training.

**Result:** The models can be evaluated on their ability to identify subscribers using recall, precision, F1-score, and ROC-AUC instead of relying only on accuracy.

## 10.2 Duplicate Records

**Challenge:** The original dataset contained 12 additional duplicate records, involving 24 rows across duplicate groups.

**Technique used:** The `drop_duplicates()` method was applied to create a cleaned dataset.

**Reason:** Exact duplicate rows can give repeated observations disproportionate influence during model training.

**Result:** Twelve additional duplicate records were removed, reducing the dataset from 41,188 to 41,176 rows.

## 10.3 Unknown Categorical Values

**Challenge:** Several categorical variables contain `unknown` values. The default-status variable contains 8,597 such records, representing approximately 20.87% of the original dataset.

**Technique used:** Unknown values were retained as explicit categories in the initial analysis. One-hot encoding was applied to categorical variables during model preprocessing.

**Reason:** Unknown values may represent unavailable information and can contain useful predictive information. Automatically replacing them could introduce unsupported assumptions.

**Result:** The categorical variables could be processed by the machine learning pipeline without discarding all records containing unknown values.

## 10.4 Data Leakage

**Challenge:** The `duration` variable represents the duration of the last contact. It is not known before a call is made and is strongly associated with the final subscription outcome.

**Technique used:** The `duration` feature was excluded from the primary predictive model.

**Reason:** Including duration in a pre-call prediction model would introduce information unavailable at the time of prediction and could lead to overly optimistic performance estimates.

**Result:** The model uses customer, campaign, and economic attributes available in the dataset, subject to verifying their availability at the intended prediction time.

## 10.5 Categorical Data Preprocessing

**Challenge:** Machine learning algorithms such as Logistic Regression require numerical inputs, while the dataset contains categorical variables such as job, education, contact method, and previous campaign outcome.

**Technique used:** `OneHotEncoder(handle_unknown="ignore")` was applied to categorical variables through a scikit-learn preprocessing pipeline.

**Reason:** One-hot encoding converts categorical values into numerical indicator variables and allows the pipeline to handle categories that were not encountered during training.

**Result:** The categorical and numerical variables could be processed consistently for model training and prediction.

## 10.6 Correlated Economic Indicators

**Challenge:** The correlation analysis identified strong positive relationships among economic indicators, including `emp.var.rate` and `euribor3m` (0.97), `euribor3m` and `nr.employed` (0.95), and `emp.var.rate` and `nr.employed` (0.91).

**Technique used:** Correlation analysis and Random Forest feature-importance analysis were conducted.

**Reason:** Highly correlated features may contain overlapping information and can complicate the interpretation of individual feature contributions.

**Result:** Economic indicators were identified as important predictive features, while their overlapping information was documented as a limitation requiring further evaluation.

## 10.7 Model Evaluation and Threshold Selection

**Challenge:** The target variable is imbalanced, and a default classification threshold may not provide the best balance between identifying subscribers and limiting unnecessary marketing contacts.

**Technique used:** Multiple classification models were compared using accuracy, precision, recall, F1-score, and ROC-AUC. The Random Forest model was also evaluated at thresholds of 0.20, 0.30, 0.40, 0.50, and 0.60.

**Reason:** Different metrics represent different aspects of model performance. Threshold analysis helps quantify the trade-off between false positives and false negatives.

**Result:** Random Forest achieved a ROC-AUC of 0.8066 and an F1-score of 0.5206 at the default threshold. The threshold comparison showed that alternative thresholds can change precision, recall, and the number of customers flagged.

## 10.8 Limitations of the Analysis

The analysis is based on historical marketing campaign records. Observed subscription rates and feature importance do not establish causal relationships.

The initial model comparison used a single stratified train-test split. Additional validation, threshold selection using training-only validation data, and evaluation on later campaign data are recommended before production deployment.

The dataset covers campaigns conducted between May 2008 and November 2010. Current customer behavior and economic conditions may differ from those historical conditions.

## 10.9 Conclusion

The project addressed data quality, categorical preprocessing, class imbalance, data leakage, correlated predictors, and model evaluation challenges.

The selected preprocessing and evaluation techniques enabled the development of a baseline classification model and comparison of multiple algorithms. Further validation is required to establish generalization performance and production suitability.
